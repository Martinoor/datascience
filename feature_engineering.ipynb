# Feature engineering (train & test)

直接调用`feature_pipeline.py`生成训练、验证和测试用的特征，保证两侧一致。

```python
import os, sys
import pandas as pd

sys.path.append(os.getcwd())
from feature_pipeline import prepare_datasets

BASE_PATH = os.getcwd()
train_path = f'{BASE_PATH}/churn-prediction-25-26/train.parquet'
test_path = f'{BASE_PATH}/churn-prediction-25-26/test.parquet'

train_df, val_df, test_df, labels, artifacts = prepare_datasets(
    train_path, test_path, val_ratio=0.2, random_state=42
)
print('train rows', len(train_df), 'val rows', len(val_df), 'test rows', len(test_df))
```

```python
print('numeric feature count', len(artifacts.numeric_cols))
print('page categories', artifacts.page_categories)
display(train_df.head())
display(test_df.head())
```

```python
# 可选：将处理好的数据缓存下来，避免重复计算
train_df.to_parquet('processed_train.parquet')
val_df.to_parquet('processed_val.parquet')
test_df.to_parquet('processed_test.parquet')
# labels.to_parquet('processed_labels.parquet')
import pickle
with open('processed_artifacts.pkl', 'wb') as f:
    pickle.dump(artifacts, f)
with open('processed_labels.pkl', 'wb') as f:
    pickle.dump(labels, f)
```
